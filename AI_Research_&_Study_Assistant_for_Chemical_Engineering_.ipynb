```python
chunk_embedding=[]
for chunk in chunks:
   embedding =genai.embed_content(
       model="models/gemini-embedding-001",
       content=chunk
   )

   chunk_embedding.append(embedding["embedding"])

print(len(chunk_embedding[0]))
```

```python
question = "What is conduction?"
```

```python
question_embedding = genai.embed_content(
    model="models/gemini-embedding-001",
    content=question
)["embedding"]
```

```python
import numpy as np
```

```python
def cosine_similarity(a, b):
    a = np.array(a)
    b = np.array(b)

    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )
```

```python
scores = []

for embedding in chunk_embedding:
    score = cosine_similarity(
        question_embedding,
        embedding
    )

    scores.append(score)

top_indices = np.argsort(scores)[-3:]

context = ""

for idx in top_indices:
    context += chunks[idx] + "\n"
```

```python
prompt = f"""
Use this context to answer:

{context}

Question:
{question}
"""
```

```python
model = genai.GenerativeModel(
    "models/gemini-2.5-flash"
)

response = model.generate_content(prompt)

print(response.text)
```
